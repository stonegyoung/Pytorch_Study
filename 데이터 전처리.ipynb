# 기본 변환
1. ToTensor()
2. Normalize(mean, std)

## ToTensor()
이미지(PIL 이미지/Numpy 배열)를 텐서 형태로 변환

[0, 255] 사이의 픽셀 값을 [0,1] 사이로 정규화(픽셀값 정규화)

```python
tensor_transform = transforms.Compose([
    transforms.ToTensor()
])
```

```python
# 0~255 사이의 픽셀값
numpy_img
```

```python
# 0~1 사이의 픽셀값
tensor_transform(numpy_img)
```

```python
# PIL에서도 적용됨
tensor_transform(img)
```

```python
# numpy: (h,w,channel)
# tensor: (channel,h,w)
numpy_img.shape, tensor_transform(img).shape
```

## Normalize(mean, std)
각 채널(RGB / 흑백)의 값들을 정규화

각 채널의 값이 평균 0, 표준편차 1을 가지도록 하여 학습을 더 안정적이고 빠르게 만들어준다.

각 채널에서 mean을 빼고 std로 나눈다

- 빨간색(R)은 평균이 0.485, 표준편차가 0.229
- 초록색(G)은 평균이 0.456, 표준편차가 0.224
- 파란색(B)은 평균이 0.406, 표준편차가 0.225
- 흑백은 평균이 0, 표준편차가 1

```python
def imgshow(array, tensor=True):
  if tensor:
    numpy_array = array.numpy()
    img = numpy_array.transpose(1,2,0)
  else:
    img = array
  plt.axis('off')
  plt.imshow(img)
```

```python
normalize_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
])
```

```python
normalize_transform(numpy_img)
```

```python
# Normalize 전
imgshow(numpy_img, tensor=False)
```

```python
# Normalize 후
imgshow(normalize_transform(numpy_img), tensor=True)
```

## ColorJitter(brightness, contrast, saturation, hue)
이미지의 밝기, 대비, 채도, 색조를 랜덤하게 변경  
변경 범위 0이 기본값
- brightness: 밝기를 변경. brightness=0.5는 밝기를 ±50% 범위 내에서 변경
- contrast: 대비를 변경. contrast=0.5는 대비를 ±50% 범위 내에서 변경
- saturation: 채도를 변경. saturation=0.5는 채도를 ±50% 범위 내에서 변경
- hue: 색조를 변경. hue=0.5는 색조를 ±50% 범위 내에서 변경

```python
colorjitter_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.ColorJitter(brightness=0.5, contrast=0.5, saturation=0.5, hue=0.5)
])

bright_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.ColorJitter(brightness=0.5)
])

contrast_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.ColorJitter(contrast=0.5)
])

saturation_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.ColorJitter(saturation=0.5)
])

hue_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.ColorJitter(hue=0.5)
])
```

## RandomResizedCrop(size, scale, ratio)
이미지를 임의로 자르고, 지정된 크기로 리사이즈

scale과 ratio로 자를 영역의 크기와 비율을 조절 -> 자른 후에는 지정된 size로 리사이즈

scale=(0.3, 0.8)은 전체 이미지에서 30%에서 80% 사이에서 자르고,  
ratio=(3/4, 4/3)은 자를 영역의 가로 세로 비율이 3:4에서 4:3 사이로 랜덤하게 선택

Resize와의 차이
- Resize: 자르기 X
- RandomResizedCrop: 자르기 있음

```python
randomresizedcrop_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.RandomResizedCrop(224, scale=(0.3,0.8), ratio=(3/4,4/3))
])
```

```python
# shape은 224로 동일
randomresizedcrop_transform(numpy_img).shape
```

```python
imgshow(randomresizedcrop_transform(numpy_img), tensor=True)
```

```python
imgshow(randomresizedcrop_transform(numpy_img), tensor=True)
```

```python
imgshow(randomresizedcrop_transform(numpy_img), tensor=True)
```
