# pivot/pivot_table

https://pandas.pydata.org/docs/user_guide/reshaping.html
- 데이터 가독성이 떨어질 때 사용
- index, column, value를 지정할 수 있다
- pivot_table은 aggfunc도 지정 가능

## pivot과 pivot_talble의 차이
- pivot_table에는 aggfunc이 있음  
-> 인덱스가 2개 이상인 경우  
-> 컬럼이 2개 이상인 경우  
-> 중복값이 있는 경우  
계산 가능

```python
df = pd.DataFrame({
    "r1": ["one", "one", "two", "three"] * 6,
    "r2": ["A", "B", "C"] * 8,
    "r3": ["foo", "foo", "foo", "bar", "bar", "bar"] * 4,
    "r4": np.random.randn(24)
})
df
```

```python
# 인덱스 두 개 이상 -> 에러
pd.pivot(df, index=['r1', 'r3'], columns='r2', values='r4')
```

```python
# 인덱스 두 개 이상
pd.pivot_table(df, index=['r1', 'r3'], columns='r2', values='r4')
```

```python
# 컬럼 두 개 이상 -> 에러
pd.pivot(df, index='r1', columns=['r2', 'r3'], values='r4')
```

```python
# 컬럼 두 개 이상. 없으면 NaN
pd.pivot_table(df, index='r1', columns=['r2', 'r3'], values='r4')
```

```python
# 중복값 있을 때 : index, columns이 같은 게 중복됨
df[(df['r1'] == 'one') & (df['r2'] == 'A')]
```

```python
# index: r1, columns: r2, values: r4 -> 에러
pd.pivot(df, index='r1', columns='r2', values='r4')
```

```python
# index: r1, columns: r2, values: r4
pd.pivot_table(df, index='r1', columns='r2', values='r4')
```

```python
# index: r1, columns: r2, values: r4 + aggfunc 시 합계로
pd.pivot_table(df, index='r1', columns='r2', values='r4', aggfunc=np.sum)
```

# stack / unstack


- 데이터 가독성이 떨어질 때 사용
stack : 컬럼 -> 인덱스 (마지막 레벨부터 인덱스로 바뀜)
unstack : 인덱스 -> 컬럼 (마지막 레벨부터 컬럼으로 바뀜)

```python
df = pd.DataFrame({
    "r1": ["one", "one", "two", "three"] * 6,
    "r2": ["A", "B", "C"] * 8,
    "r3": ["foo", "foo", "foo", "bar", "bar", "bar"] * 4,
    "r4": np.random.randn(24)
#     "r5": ["good", "bad", "soso"] * 8
})
df
```

```python
st = pd.pivot_table(df, index='r1', columns=['r2', 'r3'], values='r4')
st
```

```python
# stack 컬->인
st.stack()
```

```python
st.stack(0)
```

```python
st.stack(0).unstack(0)
```

# 데이터 병합하기
- pd.concat(): 데이터 속성이 같을 때
- pd.merge(): 공통된 항목을 기준으로 합침

## concat
- ignore_index: 기존 index 그대로/새롭게  
False면 기존(디폴트), True면 새롭게
- verify_integrity: 인덱스가 중복되는지 확인  
False면 중복 돼도 괜찮(디폴트), True면 에러 뜸
- axis  
axis=0 위아래로 합침(디폴트) axis=1 옆으로 합침  
- join: 컬럼명이 다를 때
join='outer' 합집합으로(디폴트)
join='inner' 교집합으로(동일한 컬럼만)

```python
# https://pandas.pydata.org/docs/reference/api/pandas.concat.html
s1 = pd.Series(['a', 'b'])
s2 = pd.Series(['c', 'd'])
```

```python
pd.concat([s1, s2])
```

```python
pd.concat([s1, s2], ignore_index=True)
```

```python
# verify_integrity=False 중복 돼도 괜찮
pd.concat([s1, s2], verify_integrity=False)
```

```python
# verify_integrity=True 중복되면 에러
pd.concat([s1, s2], verify_integrity=True)
```

```python
pd.concat([s1, s2], axis=1)
```

### 컬럼명이 다른 경우

```python
df1 = pd.DataFrame([['a', 1], ['b', 2]],
                   columns=['letter', 'number'])
```

```python
df2 = pd.DataFrame([['c', 3], ['d', 4]],
                   columns=['letter', 'n'])
```

```python
# 합집합으로(default임) join='outer'
pd.concat([df1, df2], join='outer')
```

```python
# 교집합으로 join='inner'
pd.concat([df1, df2], join='inner')
```

## merge
- 기준 컬럼을 가지고 병합
on = '기준 컬럼'
how = 
1. inner(디폴트): 기준 컬럼이 겹치는 것만
2. left, right: 왼쪽/오른쪽을 기준으로
3. outer: 둘 다 보이도록

```python
# https://pandas.pydata.org/docs/reference/api/pandas.merge.html
df1 = pd.DataFrame({'lkey': ['foo', 'bar', 'baz', 'foo'],
                    'value': [1, 2, 3, 5]})
df2 = pd.DataFrame({'rkey': ['foo', 'bar', 'baz', 'foo'],
                    'value': [5, 6, 7, 8]})
```

```python
pd.merge(df1, df2, on='value')
```

```python
# on 없어도 알아서 동일한 컬럼으로 해줌
pd.merge(df1, df2)
```

```python
pd.merge(df1, df2, on='value', how='left')
```

```python
pd.merge(df1, df2, on='value', how='right')
```

```python
pd.merge(df1, df2, on='value', how='outer')
```

```python
# join 기준을 인덱스를 통해
df1.set_index('value')
```

```python
df2.set_index('value')
```

```python
# left_index=True, right_index=True -> join 키로 인덱스를 사용
pd.merge(df1.set_index('value'), df2.set_index('value'), left_index=True, right_index=True)
```

```python
# join 기준 컬럼 지정하기
pd.merge(df1, df2, left_on='lkey', right_on='rkey')
```
